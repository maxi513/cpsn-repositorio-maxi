## Parametros (definicion)
Son las variables que escribimos dentro de los parentesis al crear la funcion. Son como 'espacios reservados' o etiquetas.

## Argumentos (la ejecucion)
Son los valores reales que le pasamos a la funcion cuando la llamamos.
Regla de oro: El primer argumento que envies llenara el primer parametro; el segundo argumento llenara el segundo parametro, y asi sucesivamente

Scope (alcance)
Las variables creadas dentro de una funcion son locales. esto significa que 'nacen y mueren' dentro de la funcion. si intentas usarlas afuera, el programa te dara un error porque no las conoce.

Una funcion puede hacer muchas cosas (imprimir en pantalla, calcular, mover datos), pero el return es lo que permite que el resultado salga de la funcion y pueda ser usado por el resto del programa.

return: finaliza la ejecucion de la funcion y 'entrega' un valor al lugar donde fue llamada.

capturar el valor: para usar lo que una funcion devuelve, normalmente lo guardamos en una variable.

Y si no hay return? si olvidas ponerlo, la funcion hara su trabajo pero al final le entregara un error.

```python
class Heroe:
  def __init__(self, nombre, poder, energia):
    # Atributos: datos propios del heroe
    self.nombre = nombre
    self.poder = poder
    self.energia = energia
  def presentarse(self):
    # Metodo: accion del heroe
    print(f'Soy {self.nombre}. Mi poder es {self.poder}. Tengo {self.energia}.')

  def atacar(self, villano):
    # Metodo: accion para atacar a un villano
    if self.energia >= 10:
      print(f'{self.nombre} ataca a {villano} usando {self.poder}.')
      self.energia = self.energia - 10
      print(f'Energia restante: {self.energia}')
    else:
      print(f'{self.nombre} no tiene energia suficiente para atacar.')

heroe1 = Heroe('Iron Man', 'programar armaduras', 100)
heroe2 = Heroe('Capitan America', 'Lider Avengers', 80)
heroe3 = Heroe('Hulk', 'destruir todo', 150)
heroe4 = Heroe('Batman', 'estrategista', 90)
heroe5 = Heroe('Thor', 'dios del trueno', 110)

heroe1.presentarse()
heroe1.atacar('Thanos')
```

```python
class TarjetaCredito:
  def __init__(self, banco, limite):
    self.banco = banco
    self.limite = limite

  def mostrar_limite(self):
    print(f'Tarjeta de {self.banco} con limite de ${self.limite}')

class UsuarioConTarjeta:
  def __init__(self, nombre, tarjeta):
    self.nombre = nombre
    # Aqui guardamos un objetoTarjetaCredito dentro del usuario
    self.tarjeta = tarjeta

  def mostrar_tarjeta(self):
    print(f'Usuario: {self.nombre}')
    # Usamos un metodo del objeto tarjeta.
    self.tarjeta.mostrar_limite()

mi_tarjeta = TarjetaCredito('Banco Demo', 100000)
usuario = UsuarioConTarjeta('Ana', mi_tarjeta)
usuario.mostrar_tarjeta()
```
